```python
import pandas as pd


def automated_stat_analyzer(df, column_name):

    if pd.api.types.is_numeric_dtype(df[column_name]):

        mean = df[column_name].mean()
        median = df[column_name].median()
        std = df[column_name].std()

        if mean > median:
            skewed = "Right Skewed"
        elif mean < median:
            skewed = "Left Skewed"
        else:
            skewed = "Not Skewed"

        return {
            "Mean": mean,
            "Median": median,
            "Standard Deviation": std,
            "Skewness": skewed
        }

    else:
        mode = df[column_name].mode()[0]

        return {
            "Mode": mode
        }
```

```python
import pandas as pd
import numpy as np

data = {
    'Transaction_ID': range(1, 11),
    'Product_Category': ['Electronics', 'Home', 'Electronics', 'Sports', 'Home',
                         'Electronics', 'Home', 'Sports', 'Electronics', 'Electronics'],
    'Sales_Amount': [150, 200, 155, 300, 210, 180, 205, 1000, 190, 160],
    'Customer_Age': [25, 34, np.nan, 45, 23, 31, 29, np.nan, 38, 40],
    'Rating': [5, 4, 3, 5, 2, 4, 5, 2, 4, 3]
}

df_test = pd.DataFrame(data)

print("Test dataset created successfully!")
```

```python
automated_stat_analyzer(df_test, "Sales_Amount")
```

```python
import pandas as pd


def null_handling_strategy(df, strategy="fill_mean"):

    if strategy == "drop_rows":

        return df.dropna()

    elif strategy == "fill_mean":

        df_copy = df.copy()

        numerical_columns = df_copy.select_dtypes(
            include="number"
        ).columns

        for column in numerical_columns:
            df_copy[column] = df_copy[column].fillna(
                df_copy[column].mean()
            )

        return df_copy

    elif strategy == "fill_median":

        df_copy = df.copy()

        numerical_columns = df_copy.select_dtypes(
            include="number"
        ).columns

        for column in numerical_columns:
            df_copy[column] = df_copy[column].fillna(
                df_copy[column].median()
            )

        return df_copy

    else:

        raise ValueError(
            "Invalid strategy. Use 'drop_rows', 'fill_mean', or 'fill_median'."
        )
```

```python
null_handling_strategy(df_test, "fill_mean")
```
