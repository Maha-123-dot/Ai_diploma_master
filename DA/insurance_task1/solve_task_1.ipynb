```python
import pandas as pd
from sklearn.preprocessing import StandardScaler
```

```python
# Load the dataset
df = pd.read_csv("insurance.csv")

print("Dataset shape:", df.shape)
display(df.head())
```

```python
# Check dataset information
print("Dataset information:")
df.info()

print("\nMissing values:")
print(df.isnull().sum())
```

```python
# Separate features (X) and target (Y)
X = df.drop("charges", axis=1)
Y = df["charges"]

print("X columns:", list(X.columns))
print("Y column: charges")
```

```python
# Convert categorical columns to numerical values
X = pd.get_dummies(
    X,
    columns=["sex", "smoker", "region"],
    drop_first=True,
    dtype=int
)

display(X.head())
```

```python
# Scale X
scaler = StandardScaler()
X = pd.DataFrame(
    scaler.fit_transform(X),
    columns=X.columns
)

# Make sure Y is numerical
Y = pd.to_numeric(Y)

print("Preprocessed X:")
display(X.head())

print("Y:")
display(Y.head())
```

```python
print("Final X data types:")
print(X.dtypes)

print("\nFinal Y data type:")
print(Y.dtype)

print("\nAre all X columns numerical?",
      X.select_dtypes(include="number").shape[1] == X.shape[1])

print("Is Y numerical?",
      pd.api.types.is_numeric_dtype(Y))
```
