```python
from abc import ABC, abstractmethod 
class Animal (ABC):
    
    @abstractmethod
    def move(self):
        pass

    class Bird(Animal):
        def move(self):
            print("move from bird")
    class cat(Animal):
        def move(self):
            print("move from cat")
 
```

```python
a=Bird()
a.move()
```

```python
class Employee:
    def __init__(self, name, salary):
        self. name=name
        self.salary=salary
    def set_salary(self,obj,new_sal):
        obj.salary=new_sal 
class Manager(Employee):
    def __init__(self, name, salary):
        self.name=name
        self.salary=salary
        
```

```python
E1=Employee("Depi", 2000)
m1=Manager("Maha",5000)
E1.set_salary(m1, 6000)
print(m1.salary)
```
