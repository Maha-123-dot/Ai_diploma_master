```python
class Table:
    def __init__(self, number):
        self.number = number

    def print_table(self):
        for i in range(1, 11):
            print(self.number * i)


x = Table(9)
x.print_table()
```

```python
class Prime:
    def is_prime(self, number):
        if number < 2:
            return False

        for i in range(2, number):
            if number % i == 0:
                return False

        return True

    def twin_prime(self):
        for i in range(2, 100):
            if self.is_prime(i) and self.is_prime(i + 2):
                print(i, i + 2)


x = Prime()
x.twin_prime()
```

```python
class Factors:
    def __init__(self, number):
        self.number = number

    def prime_factors(self):
        i = 2

        while self.number > 1:
            if self.number % i == 0:
                print(i)
                self.number = self.number // i
            else:
                i += 1


x = Factors(56)
x.prime_factors()
```

```python
class Binary:
    def __init__(self, number):
        self.number = number

    def convert(self):
        binary = ""

        while self.number > 0:
            binary = str(self.number % 2) + binary
            self.number = self.number // 2

        return binary


x = Binary(10)
print(x.convert())
```

```python
class Perfect:
    def check(self, number):
        total = 0

        for i in range(1, number):
            if number % i == 0:
                total += i

        if total == number:
            print(number)

    def perfect_numbers(self):
        for i in range(1, 1000):
            self.check(i)


x = Perfect()
x.perfect_numbers()
```

```python
class Calculator:
    def add(self, a, b):
        return a + b

    def subtract(self, a, b):
        return a - b

    def multiply(self, a, b):
        return a * b

    def divide(self, a, b):
        return a / b


x = Calculator()

print(x.add(10, 5))
print(x.subtract(10, 5))
print(x.multiply(10, 5))
print(x.divide(10, 5))
```
